```python
with open("facts.txt", "r") as f:
    raw_facts = [line.strip() for line in f.readlines()]
with open("rules.txt", "r") as f:
    raw_rules = [line.strip() for line in f.readlines()]
```

```python
rules = []
facts = []
for rule in raw_rules:
    rule = rule.replace("IF", "")
    rule = rule.replace("is", "=")
    rule = rule.replace("AND", "&")
    rule = rule.replace("OR", "|")
    rules.append(rule.split("THEN"))
for fact in raw_facts:
    fact = fact.replace("is", "=")
    facts.append(fact)
print(rules)
print(facts)
```

```python
processed_rules = []
for rule in rules:
    conditions = rule[0]
    conclusion = rule[1]
    if "&" in conditions:
        conditions = conditions.split("&")
        connector = "%"
        conditions = [condition.strip() for condition in conditions]
    elif "|" in conditions:
        conditions = conditions.split("|")
        connector = "|"
        conditions = [condition.strip() for condition in conditions]
    else:
        conditions = [conditions.strip()]
        connector = "_"
    processed_conditions = []
    for condition in conditions:
        if len(condition.split("=")) == 2:
            variable = condition.split("=")[0].strip()
            value = condition.split("=")[1].strip()
            condition = (variable, value)
        elif len(condition.split(">")) == 2:
            variable = condition.split(">")[0].strip()
            value = condition.split(">")[1].strip()
            condition = (variable, ">", value)
        elif len(condition.split("<")) == 2:
            variable = condition.split("<")[0].strip()
            value = condition.split("<")[1].strip()
            condition = (variable, "<", value)
        else:
            variable = condition.strip()
            condition = (variable, True)
        processed_conditions.append(condition)
    processed_rules.append((processed_conditions, conclusion, connector))
processed_rules
```

```python
processed_facts = []
for fact in facts:
    if len(fact.split("=")) == 2:
        variable = fact.split("=")[0].strip()
        value = fact.split("=")[1].strip()
        fact = (variable, value)
    else:
        variable = fact.strip()
        fact = (variable, True)
    processed_facts.append(fact)
processed_facts
```
